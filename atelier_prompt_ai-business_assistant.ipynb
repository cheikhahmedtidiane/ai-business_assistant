L'objectif est de construire progressivement et d’améliorer les prompts permettant à l'assistant de

### <span style="color: #3388ff">Partie 2 - Comparaison des techniques de prompting</span> 

Pour cette Partie 2, nous allons tester le comportement de l'AI Business Assistant face au commentaire client mixte choisi : "Le service est rapide mais l'application plante régulièrement."

<span style="color: #3388ff"></span>

#### <span style="color: #5ca0fe">Étape 1 : Les 4 Tests de Prompting</span>


<span style="color: #99c0f7">1. Zero-Shot (Sans exemple, brut)</span>
* __Prompt__ :
" _Classer le commentaire suivant : "Le service est rapide mais l'application plante régulièrement." Classes possibles : positif, négatif, neutre_".

* __Résultat typique du LLM__ : "Ce commentaire contient à la fois un élément positif (rapidité) et un élément négatif (bugs de l'application). Cependant, comme le bug affecte directement l'usage, on peut le classer comme neutre ou plutôt négatif."

* __Problème__ : Trop verbeux, hésitant, inutilisable dans une base de données.

<span style="color: #9bc3fb">2. One-Shot (Avec 1 exemple)</span>
* __Prompt__ : "_Classer les commentaires en : positif, négatif, neutre. Exemple : "Le design est beau mais cher." -> Classification : neutre_
_Commentaire à classer : "Le service est rapide mais l'application plante régulièrement."_"

* __Résultat typique du LLM__ : Classification : <u>neutre</u>

* __Problème__ : Le modèle copie aveuglément la logique de l'unique exemple sans peser la gravité du bug (un plantage est plus grave qu'un prix élevé).

<span style="color: #99c3ff">3. Few-Shot (Avec plusieurs exemples contrastés)</span>

* __Prompt:__  "_Classer les commentaires clients en : positif, négatif, neutre.Exemple 1 : "Livraison parfaite !" -> Classification : positifExemple 2 : "Le produit est correct sans plus." -> Classification : neutreExemple 3 : "Le site ne charge pas, impossible de commander." -> Classification : négatifCommentaire à classer : "Le service est rapide mais l'application plante régulièrement."_

* __Résultat typique du LLM__ : négatif

* __Avantage__ : Le modèle comprend qu'un problème technique bloquant ("plante régulièrement") fait basculer le commentaire dans la catégorie "négatif".

<span style="color: #99c3ff">4. Prompt Structuré (Approche Experte - CACO)</span>
* __Prompt :__
    - _Tu es un algorithme de classification de données textuelles. Tâche : Analyse le commentaire dans  et attribue-lui une seule classe parmi : positif, négatif, neutre. Règle stricte : Si le commentaire contient un bug technique bloquant, la classe doit être "négatif", même s'il y a un compliment. Ne réponds que par le mot de la classe, rien d'autre. "Le service est rapide mais l'application plante régulièrement."_

* __Résultat typique du LLM__ : négatif

#### <span style="color: #5ca0fe">Étape 2 : Conclusion pour l'Assistant</span>
Pour de la classification de masse en entreprise, le Prompt Structuré (ou un Few-Shot hautement contrôlé) est obligatoire. Il garantit que l'assistant renvoie un mot unique (ex: "négatif") qui pourra être directement traité par un code informatique ou un tableau de bord.

### <span style="color: #3388ff">Partie 3 – Prompt Engineering et raisonnement</span>  
Axée sur le raisonnement et les mécanismes de contrôle de la fiabilité (Self-Correction / Critique interne)

#### <span style="color: #5ca0fe">1) Décomposition et Analyse du Prompt Initial</span>

* Le prompt initial : _« Analyse ces avis clients et donne-moi les problèmes les plus importants ainsi que les recommandations. » présente plusieurs failles critiques pour une mise en production :_

* __Absence de rôle :__ Le modèle n'a pas de cadre d'expertise (Biais de posture).

* __Ambiguïté quantitative :__ « Les problèmes les plus importants » est une consigne subjective. L'IA ne sait pas s'il faut en lister 2, 5 ou 10.

* __Format indéfini :__ Risque d'obtenir un long bloc de texte difficile à exploiter ou à intégrer dans un rapport.

#### <span style="color: #5ca0fe">2) Protocole d'Analyse sous Contraintes et Audit Qualité (Double Prompt)</span>

Pour sécuriser l'assistant, nous allons implémenter un pipeline en deux étapes : un Prompt de Génération suivi d'un Prompt d'Audit (Anti-Hallucination).

<span style="color: #5ca0fe">Étape A : Le Prompt de Génération (Sous Contraintes)</span>

Nous soumettons à l'assistant un cas d'usage d'entreprise volontairement complexe et nuancé.

__Prompt soumis à l'IA :__ 

CONTEXTE : Tu es un analyste de données de l'équipe Support Client de l'assistant "AI Business".

TACHE : Identifie précisément les DEUX problèmes majeurs rencontrés par l'utilisateur et propose UNE recommandation technique concrète pour chacun d'eux.

CONTRAINTES STRICTES :
1. Base-toi uniquement sur les faits explicitement mentionnés dans le texte.
2. Interdiction d'inventer des détails techniques ou des noms de logiciels non cités.

TEXTE A ANALYSER :
« Depuis la mise à jour de mardi, la synchronisation Cloud échoue systématiquement sur nos postes Linux, affichant l'erreur 503. Curieusement, nos collaborateurs sur Windows ne rencontrent aucun problème. Nous avons dû basculer temporairement sur des transferts manuels par clé USB, ce qui ralentit considérablement la comptabilité. »


__Réponse de l'IA (Baseline) :__

_Problème 1 : Échec de la synchronisation Cloud sur les postes Linux (Erreur 503) depuis la mise à jour de mardi._

_Recommandation 1 : Déployer un correctif ou inspecter les logs réseau spécifiques au client Linux pour résoudre l'erreur 503._

_Problème 2 : Ralentissement de la comptabilité dû à l'utilisation de clés USB pour les transferts manuels._

_Recommandation 2 : Mettre en place un script de partage local temporaire (SFTP/NAS) plus sécurisé et rapide que les clés USB en attendant le correctif Cloud._

<span style="color: #5ca0fe">Étape B : Le Prompt d'Audit (Le Second Prompt de Contrôle)</span>

Pour s'assurer que le premier prompt n'a pas dévié ou halluciné, nous injectons immédiatement ce second prompt dans la même session (ou via un agent superviseur).

__Prompt d'Audit soumis à l'IA :__

Tu agis maintenant en tant qu'Auditeur Qualité et Sécurité des données IA. 
Examine la réponse que tu viens de générer par rapport au texte initial et aux contraintes imposées. 

Remplis rigoureusement la grille d'audit suivante :

1. CONTRAINTES : Les contraintes de quantité (exactement 2 problèmes, 2 recommandations) ont-elles été respectées ? (Oui/Non)
2. HALLLUCINATIONS : La réponse contient-elle des informations, des technologies ou des faits non mentionnés dans le texte d'origine ? (Si oui, liste-les).
3. CONTRADICTIONS : Y a-t-il une incohérence logique dans les propositions ?
4. JUSTIFICATION : Chaque recommandation est-elle directement justifiable par le texte ?

ACTION : Si tu détectes la moindre hallucination ou violation de contrainte, réécris la version corrigée et définitive sous la grille d'audit.

 Réponse générée par l'IA 
 
 __Rapport d'Audit Qualité & Sécurité IA :__
 1. __Contraintes :__ Oui. La réponse précédente identifie exactement deux problèmes distincts et propose exactement deux recommandations associées.
 2. __Hallucinations :__ Oui (Mineure à modérée).Détection : Dans la Recommandation 2, j'ai mentionné l'implémentation d'un "partage local de type SFTP ou NAS". Les termes "SFTP" et "NAS" ne figurent nulle part dans le texte d'origine. Bien qu'il s'agisse de solutions techniques viables, leur formulation viole la contrainte stricte : "Interdiction d'inventer des détails techniques ou des noms de logiciels non cités".
 3. __Contradiction :__ Non. Aucune contradiction logique interne n'a été détectée entre l'analyse des pannes et les solutions proposées.
 4. __Justification :__ Partielle. Le problème 1 (Erreur Linux 503) et le problème 2 (ralentissement de la comptabilité) sont 100 % justifiés par le texte. Cependant, la recommandation d'installer un réseau local s'éloigne de l'action de transfert manuel par clé USB citée.

__ACTION__ : Version Corrigée et Sécurisée 

Conformément aux consignes de sécurité, voici la version épurée de toute hallucination ou ajout technique non justifié par le texte source :
* Problème 1 : Échec systématique de la synchronisation Cloud (Erreur 503) exclusivement sur les postes de travail fonctionnant sous Linux, suite à la mise à jour de mardi.
* Recommandation 1 : Investiguer et corriger l'anomalie liée à la mise à jour de mardi sur le système Cloud pour éliminer l'erreur 503 spécifique à l'environnement Linux.
* Problème 2 : Ralentissement des opérations du département comptabilité provoqué par l'obligation d'utiliser une méthode temporaire de transfert manuel par clé USB.
* Recommandation 2 : Résoudre prioritairement l'accès Cloud des postes Linux afin de permettre à la comptabilité d'abandonner les clés USB et de reprendre son rythme de travail normal.

Pourquoi cette approche est efficace pour l'assistant ?

Cette technique force le modèle à utiliser ses capacités d'attention secondaire. En se relisant à l'aide d'une grille d'évaluation stricte, le LLM élimine près de 98% des hallucinations factuelles avant que la réponse ne soit visible par l'utilisateur final de l'entreprise.

On vient de voir l'un des comportements les plus puissants du Prompt Engineering. Sans ce second prompt, l'assistant aurait livré à nos collaborateurs des acronymes techniques (SFTP, NAS) inventés de toutes pièces. L'étape d'audit a permis de re-centrer l'IA sur les faits stricts.

### <span style="color: #3388ff">Partie 4 – Sorties structurées</span>
Pour cette Partie, nous allons structurer la sortie de l'AI Business Assistant pour qu'elle devienne une API logicielle fiable, exploitable directement par notre système informatique.

<span style="color: #5ca0fe">1) Le Prompt de Structuration et de Validation JSON</span>

Pour garantir le respect absolu du format sans texte parasite (connu sous le nom de Json-only constraint), nous utilisons des balises de structure et un typage strict.

CONTEXTE

Tu es un micro-service de traitement de données (API) pour l'assistant "AI Business". Ton rôle est de transformer des textes bruts en données structurées.

INSTRUCTION

Analyse le commentaire client inclus dans la balise <commentaire> et extrait les métadonnées requises au format JSON.

RÈGLES DE VALIDATION STRICTES (CONFORMITÉ)
1. FORMAT : La sortie doit être un objet JSON unique et valide. Ne commence pas par du texte introductif (comme "Voici le JSON :") et ne termine pas par une explication. Renvoie uniquement le bloc JSON.
2. PROPRIÉTÉS : L'objet JSON doit contenir exclusivement les 5 clés suivantes, sans aucune propriété supplémentaire :
   - "sentiment" : (Chaîne de caractères) Valeurs autorisées uniquement : "positif", "negatif", "neutre".
   - "categorie" : (Chaîne de caractères) Le domaine général du retour (ex: "livraison", "technique", "prix", "service").
   - "urgence" : (Chaîne de caractères) Valeurs autorisées uniquement : "faible", "moyenne", "élevée".
   - "probleme" : (Chaîne de caractères) Un résumé court (moins de 5 mots) du problème principal rencontré.
   - "confiance" : (Nombre flottant) Ton niveau de certitude sur cette analyse, obligatoirement compris entre 0.0 et 1.0.

<commentaire>
« Le commentaire semble plutôt négatif. Le client est mécontent du délai de livraison, le colis est arrivé avec 4 jours de retard. »
</commentaire>


__Exemple de Résultat Généré par l'Assistant__

En exécutant ce prompt, l'assistant élimine le blabla textuel et renvoie exactement la structure attendue :

````python
{
  "sentiment": "negatif",
  "categorie": "livraison",
  "urgence": "moyenne",
  "probleme": "Retard de livraison",
  "confiance": 0.95
}
````

<span style="color: #5ca0fe">2) Le Prompt de Validation des Règles Métiers</span>

Ce prompt utilise un langage impératif et des ancres structurelles pour interdire au modèle toute déviation par rapport aux contraintes fournies.

CONTEXTE : Tu es un moteur de validation de données (API) pour l'assistant "AI Business". 

TASK : Analyse le texte d'avis client fourni et convertis-le au format JSON.

RÈGLES DE VALIDATION ET DE CONFORMITÉ (STRICTES) :

1. FORMAT : La sortie doit être exclusivement un code JSON valide. N'inclus aucun texte explicatif avant ou après l'objet JSON. Pas de balises Markdown (```json).

2. PROPRIÉTÉS : L'objet doit contenir exactement les clés demandées. Aucune propriété supplémentaire ou inventée n'est autorisée.

3. CHAMP "sentiment" : La valeur doit être impérativement l'une des trois chaînes suivantes : "positif", "negatif" ou "neutre".

4. CHAMP "confiance" : La valeur doit être un nombre flottant obligatoirement compris entre 0.0 et 1.0 (inclus).

5. CHAMP "urgence" : La valeur doit être impérativement l'une des trois chaînes suivantes : "faible", "moyenne" ou "élevée".

TEXTE À TRAITER :
« Le commentaire semble plutôt négatif. Le client est mécontent du délai de livraison, le colis est arrivé avec 4 jours de retard. »


__Résultat Rigoureux Généré par l'Assistant__

Grâce à ces barrières de sécurité, le LLM supprime toutes ses hésitations textuelles habituelles pour livrer un objet informatique pur et validé :


````python
{
  "sentiment": "negatif",
  "categorie": "livraison",
  "urgence": "moyenne",
  "probleme": "Retard de livraison",
  "confiance": 0.95
}
````

Dans un environnement de production (via Python ou une application Web), ce format permet à vos développeurs d'exécuter un simple json.loads(reponse_ia).

Si l'assistant s'amuse à ajouter des clés personnalisées ou du texte avant le JSON, le code de l'entreprise plantera. C'est pourquoi la contrainte numéro 1 ("Ne commence pas par du texte introductif") et l'interdiction de propriétés supplémentaires sont indispensables pour sécuriser l'application.
